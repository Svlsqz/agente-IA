```python
import json

filename = "entrenamiento_casa_elena.jsonl"

with open(filename, "r", encoding="utf-8-sig") as file:
    for line in file:
        try:
            json.loads(line)
        except json.JSONDecodeError as e:
            print(f"Error en la línea: {line}")
            print(f"Detalle del error: {e}")
```
